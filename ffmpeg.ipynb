```python
import os
from datetime import datetime
import pytz

# 1. Mount Drive (jika belum)
from google.colab import drive
drive.mount('/content/drive')

# 2. Tentukan file yang ingin digabungkan
# @markdown ### Masukkan alamat lengkap file audio dari Google Drive, satu per baris.
# @markdown (Tips: Klik kanan file di sidebar kiri Drive -> Copy Path)
audio_files_input_raw = """isi disini""" #

# Split the input string into a list of file paths, filtering out empty lines
files_to_join = [f.strip() for f in audio_files_input_raw.split('\n') if f.strip()]

if not files_to_join:
    print("Tidak ada file yang dimasukkan untuk digabungkan.")
else:
    # 3. Buat file list.txt secara otomatis
    with open("list.txt", "w") as f:
        for file in files_to_join:
            f.write(f"file '{file}'\n")

    # 4. Ambil waktu sekarang untuk nama file
    zona_wib = pytz.timezone('Asia/Jakarta')
    waktu_sekarang = datetime.now(zona_wib).strftime("%d-%m-%Y_%H-%M")
    nama_output = f"/content/drive/MyDrive/for_ffmpeg/bahanaudio/hasil_penyatuan_{waktu_sekarang}.mp3"

    print(f"Menggabungkan {len(files_to_join)} file audio...")
    # 5. Jalankan FFmpeg
    os.system(f'ffmpeg -f concat -safe 0 -i list.txt -c copy "{nama_output}"')

    print(f"Selesai! File disimpan dengan nama: {nama_output}")
```

```python
import os
from datetime import datetime
import pytz

# 2. Tentukan file yang ingin digabungkan
# @markdown ### Masukkan alamat lengkap file video dari Google Drive, satu per baris.
# @markdown (Tips: Klik kanan file di sidebar kiri Drive -> Copy Path)
video_files_input_raw = """isi disini""" #

# Split the input string into a list of file paths, filtering out empty lines
files_to_join = [f.strip() for f in video_files_input_raw.split('\n') if f.strip()]

if not files_to_join:
    print("Tidak ada file video yang dimasukkan untuk digabungkan.")
else:
    # 3. Buat file list.txt secara otomatis
    # Gunakan 'file:' prefix untuk setiap path
    with open("list_video.txt", "w") as f:
        for file in files_to_join:
            f.write(f"file '{file}'\n")

    # 4. Ambil waktu sekarang untuk nama file
    zona_wib = pytz.timezone('Asia/Jakarta')
    waktu_sekarang = datetime.now(zona_wib).strftime("%Y%m%d_%H%M%S")

    # Lokasi simpan seperti yang diminta user
    folder_simpan = "/content/drive/MyDrive/for_ffmpeg/bahanvideo/"
    if not os.path.exists(folder_simpan):
        os.makedirs(folder_simpan)

    nama_output = f"hasil_penyatuan_{waktu_sekarang}.mp4"
    path_output_final = os.path.join(folder_simpan, nama_output)

    print(f"Menggabungkan {len(files_to_join)} file video...")
    print(f"Output akan disimpan di: {path_output_final}")

    # 5. Jalankan FFmpeg untuk menggabungkan video tanpa re-encoding
    # -f concat: Menggunakan demuxer concat
    # -safe 0: Mengizinkan path absolut atau relatif di list.txt (penting untuk Drive)
    # -c copy: Menyalin stream tanpa re-encoding (cepat dan menjaga kualitas)
    try:
        # tambahkan map 0 jika ingin semua track nya ikut, jika tidak maka -map 0:v -map 0:a
        get_ipython().system(f'ffmpeg -f concat -safe 0 -i list_video.txt -c copy "{path_output_final}"')
        print(f"\nSelesai! Video gabungan sudah tersimpan di: {path_output_final}")
    except Exception as e:
        print(f"\n[ERROR] Terjadi kesalahan saat menjalankan FFmpeg: {e}")
        print("Pastikan semua file video yang dimasukkan valid dan dapat diakses.")
```

```python
import os
from google.colab import drive

# @title Memotong Video Panjang menjadi Pendek
# @markdown ---
# @markdown ## Langkah 1: Masukkan Alamat File Video dari Google Drive
# @markdown (Tips: Klik kanan file di sidebar kiri Drive -> Copy Path)
input_file_path = "" # @param {type:"string"}

# Memastikan Google Drive terhubung jika belum
if not os.path.exists('/content/drive'):
    print("Menghubungkan Google Drive...")
    drive.mount('/content/drive')
else:
    print("Google Drive sudah terhubung.")

# Cek apakah file input_file_path ada
if not os.path.exists(input_file_path):
    print(f"[ERROR] File '{input_file_path}' tidak ditemukan di Google Drive Anda. Mohon periksa kembali alamatnya.")
else:
    print(f"File '{input_file_path}' ditemukan.")
    input_file_name = input_file_path # Menggunakan nama file lengkap sebagai input untuk FFmpeg

    # @markdown ---
    # @markdown ## Langkah 2: Atur Durasi Setiap Potongan
    # @markdown Masukkan durasi (dalam detik) untuk setiap segmen video yang ingin Anda potong.
    # @markdown Contoh: Untuk memotong video menjadi segmen 30 menit, masukkan `1800` (30 * 60).
    segment_duration_str = "" # @param {type:"string"}
    try:
        detik = int(segment_duration_str)
        if detik <= 0:
            raise ValueError("Durasi harus angka positif.")
    except ValueError as e:
        print(f"[ERROR] Input durasi tidak valid: {e}. Menggunakan durasi default 1800 detik.")
        detik = 1800

    # @markdown ---
    # @markdown ## Langkah 3: Tentukan Lokasi Penyimpanan Output
    # @markdown Pilih apakah Anda ingin menyimpan video hasil potongan ke Google Drive atau di penyimpanan sementara Colab.
    # @markdown **Penting:** Jika memilih Google Drive, pastikan Anda telah menjalankan dan mengotentikasi sel `drive.mount('/content/drive')` sebelumnya.
    save_to_google_drive = True # @param {type:"boolean"}
    google_drive_output_folder = "" # @param {type:"string"}

    output_folder = "/content/video_segments_temp/" # Default to Colab storage

    if save_to_google_drive:
        if os.path.exists('/content/drive'):
            output_folder = google_drive_output_folder
            if not os.path.exists(output_folder):
                os.makedirs(output_folder, exist_ok=True)
            print(f"Output akan disimpan di Google Drive: {output_folder}")
        else:
            print("[PERINGATAN] Google Drive belum terpasang.")
            print("Anda perlu menjalankan sel 'drive.mount('/content/drive')' terlebih dahulu untuk menyimpan ke Drive.")
            print("Output akan disimpan di penyimpanan sementara Colab sebagai gantinya.")
            save_to_google_drive = False # Revert to local save
            if not os.path.exists(output_folder):
                os.makedirs(output_folder, exist_ok=True)
    else:
        if not os.path.exists(output_folder):
            os.makedirs(output_folder, exist_ok=True)
        print("Output akan disimpan di penyimpanan sementara Colab.")

    # Extract filename without extension and get original extension
    # For Drive path, base_name will be the actual file name, not the full path
    base_name = os.path.splitext(os.path.basename(input_file_name))[0]
    file_extension = os.path.splitext(input_file_name)[1]

    # Construct FFmpeg command
    output_pattern = os.path.join(output_folder, f"{base_name}_PART_%03d.mkv")

    print(f"\nMemulai proses pemotongan video '{os.path.basename(input_file_name)}' setiap {detik} detik...")

    # tambahkan map 0 jika ingin semua track nya ikut, jika tidak maka -map 0:v -map 0:a
    ffmpeg_command = f'ffmpeg -i "{input_file_name}" -f segment -segment_time {detik} -reset_timestamps 1 -map 0 -c:v copy -c:a copy "{output_pattern}"'

    # Execute FFmpeg
    try:
        get_ipython().system(ffmpeg_command)
        print("\nProses pemotongan video selesai!")
        print(f"File-file hasil potongan tersedia di folder '{output_folder}'.")
        print("\nDaftar file yang dihasilkan:")
        get_ipython().system(f'ls -lh "{output_folder}"')
    except Exception as e:
        print(f"\n[ERROR] Terjadi kesalahan saat menjalankan FFmpeg: {e}")
        print("Pastikan file video yang dimasukkan valid dan FFmpeg terinstal dengan benar.")
```

```python
import os
from google.colab import drive

# @title Batch Konversi MKV ke MP4
# @markdown *Untuk banyak file sekaligus*
# @markdown ### 1. Masukkan Alamat FOLDER MKV
# @markdown (Masukkan path folder yang berisi banyak file MKV, contoh: `/content/drive/MyDrive/video_mkv/`)
input_folder_path = "" # @param {type:"string"}

# @markdown ---
# @markdown ### 2. Tentukan Lokasi Penyimpanan Output
save_to_google_drive = True # @param {type:"boolean"}
google_drive_output_folder = "" # @param {type:"string"}

output_folder_colab = "/content/converted_videos/"

# Tentukan folder output akhir
if save_to_google_drive:
    if not os.path.exists('/content/drive'):
        print("[PERINGATAN] Menghubungkan Google Drive...")
        drive.mount('/content/drive')

    output_folder = google_drive_output_folder
else:
    output_folder = output_folder_colab

if not os.path.exists(output_folder):
    os.makedirs(output_folder, exist_ok=True)

# Proses Batch
if not os.path.exists(input_folder_path):
    print(f"[ERROR] Folder input '{input_folder_path}' tidak ditemukan.")
else:
    # List semua file di folder input
    all_files = os.listdir(input_folder_path)
    # Filter hanya file .mkv
    mkv_files = [f for f in all_files if f.lower().endswith('.mkv')]

    if not mkv_files:
        print("Tidak ditemukan file .mkv di folder tersebut.")
    else:
        print(f"Ditemukan {len(mkv_files)} file. Memulai konversi...\n")

        for index, filename in enumerate(mkv_files, 1):
            input_mkv_path = os.path.join(input_folder_path, filename)
            base_name = os.path.splitext(filename)[0]
            output_mp4_path = os.path.join(output_folder, f"{base_name}.mp4")

            print(f"[{index}/{len(mkv_files)}] Memproses: {filename}")

            # FFmpeg command (copy stream agar cepat & tanpa kehilangan kualitas)
            ffmpeg_command = f'ffmpeg -i "{input_mkv_path}" -c:v copy -c:a copy -loglevel error "{output_mp4_path}"'

            try:
                # Jalankan perintah
                get_ipython().system(ffmpeg_command)
                print(f"--- Selesai: {base_name}.mp4")
            except Exception as e:
                print(f"--- [ERROR] Gagal memproses {filename}: {e}")

        print("\n" + "="*30)
        print("SEMUA PROSES SELESAI!")
        print(f"Hasil tersimpan di: {output_folder}")
        get_ipython().system(f'ls -lh "{output_folder}"')
```

```python
import os
import subprocess
from google.colab import drive

# @title 🔊 Pengeras Suara Video ke Google Drive
# @markdown Masukkan path file video yang ada di Google Drive atau yang diunggah ke Colab.

# @markdown ---
# @markdown ### Pengaturan Path
input_path = "" #@param {type:"string"}
drive_folder = "" #@param {type:"string"}

# --- Proses Persiapan ---

# 1. Mount Google Drive
if not os.path.exists('/content/drive'):
    drive.mount('/content/drive')

# 2. Logika Penamaan File
# Mengambil nama file tanpa ekstensi dan ekstensinya
base_name = os.path.basename(input_path)
file_name, file_ext = os.path.splitext(base_name)

# Format: nama-input_normalized
output_name = f"{file_name}_normalized{file_ext}"

# 3. Menyiapkan Folder Tujuan di Drive
# Perbaikan: Langsung gunakan drive_folder karena sudah path lengkap
target_dir = drive_folder
os.makedirs(target_dir, exist_ok=True)

output_path = os.path.join(target_dir, output_name)

# 4. Eksekusi FFmpeg
if os.path.exists(input_path):
    print(f"File ditemukan: {base_name}")
    print(f"Memproses ke: {output_path}...")

# @markdown ---
# @markdown ### 🔊 Pilih Metode Volume

metode = "volume multiplier (kelipatan)" # @param ["loudnorm (Auto Normalize)", "volume multiplier (kelipatan)", "volume dB (desibel)"]

# @markdown ### 📐 Nilai Pengaturan (aktif sesuai metode dipilih)
volume_multiplier = 2.0 # @param {type:"number"}
# @markdown *(hanya aktif jika metode = volume multiplier, contoh: 2.0 = 2x lipat)*

volume_db = 0 # @param {type:"integer"}
# @markdown *(hanya aktif jika metode = volume dB, contoh: 10 = tambah 10dB)*

# ── Logic pemilihan metode ──────────────────────────────────────
if metode == "loudnorm (Auto Normalize)":
    af_filter = "loudnorm"
    print(f"✅ Metode: loudnorm (normalisasi otomatis)")

elif metode == "volume multiplier (kelipatan)":
    af_filter = f"volume={volume_multiplier}"
    print(f"✅ Metode: volume multiplier → {volume_multiplier}x lipat")

elif metode == "volume dB (desibel)":
    af_filter = f"volume={volume_db}dB"
    print(f"✅ Metode: volume dB → +{volume_db} dB")

# ── Jalankan FFmpeg ─────────────────────────────────────────────
cmd = ["ffmpeg", "-i", input_path, "-af", af_filter, "-c:v", "copy", output_path, "-y"]

print(f"\n🚀 Menjalankan: {' '.join(cmd)}\n")
subprocess.run(cmd, check=True)
print(f"\n✅ Selesai! Output: {output_path}")

if os.path.exists(output_path):
    # Kondisi ideal: Output ada (berarti proses sukses)
    print(f"\n✅ BERHASIL!")
    print(f"File tersimpan di Drive: Drive Saya/{drive_folder}/{output_name}")

elif not os.path.exists(input_path):
    # Kondisi parah: Ternyata file inputnya yang dari awal memang tidak ada
    print(f"❌ Error: File INPUT '{input_path}' tidak ditemukan.")

else:
    # Kondisi tengah: Input ada, tapi output tidak muncul (berarti prosesnya gagal/crash)
    print(f"\n❌ ERROR: Input ditemukan, tapi gagal memproses OUTPUT ke '{output_path}'.")
```

```python
# @title Kirim file dari Google Drive ke Tele
# @markdown *Untuk banyak file sekaligus*

# 1. Instalasi Library
!pip install pyrogram tgcrypto

import os
import asyncio
import pytz
from datetime import datetime
from pyrogram import Client

# 2. Masukkan Data API
api_id = ""
api_hash = ""
# Masukkan path FOLDER-nya di sini
folder_path = "" # @param{type:"string"}

async def upload_files():
    async with Client("my_session", api_id, api_hash) as app:
        # List semua file di dalam folder
        files = [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]

        print(f"📂 Ditemukan {len(files)} file. Memulai proses upload...\n")

        for index, file_name in enumerate(files, 1):
            file_path = os.path.join(folder_path, file_name)

            # --- Bagian Penentu Caption Otomatis ---
            ukuran_byte = os.path.getsize(file_path)
            ukuran_gb = ukuran_byte / (1024**3)
            zona_wib = pytz.timezone('Asia/Jakarta')
            waktu_sekarang = datetime.now(zona_wib).strftime("%d-%m-%Y %H:%M")

            caption_keren = (
                f"📄 **Nama ({index}/{len(files)}):** `{file_name}`\n"
                f"📦 **Ukuran:** {ukuran_gb:.2f} GB\n"
                f"⏰ **Selesai:** {waktu_sekarang}\n"
                f"🚀 *Dikirim otomatis via Google Colab*"
            )
            # ---------------------------------------

            print(f"🚀 [{index}/{len(files)}] Memulai upload: {file_name}...")

            try:
                await app.send_document(
                    chat_id="me",
                    document=file_path,
                    caption=caption_keren,
                    progress=progress
                )
                print(f"\n✅ {file_name} terkirim!")

                # Jeda singkat agar tidak terkena FloodWait (Limit) dari Telegram
                await asyncio.sleep(2)

            except Exception as e:
                print(f"\n❌ Gagal mengirim {file_name}: {e}")

def progress(current, total):
    print(f"📤 Uploading: {current * 100 / total:.1f}%", end="\r")

# 3. Menjalankan fungsi
await upload_files()
```

```python
#@title 🚀 Alur Kerja Video: Transkrip Potong Blurlatar
#@markdown Masukkan konfigurasi dan jalankan sel ini.

# Instal
try:
    import whisper
except ImportError:
    !pip install openai-whisper

import subprocess
import os
from tqdm.notebook import tqdm

# --- CONFIGURASI PARAMETER ---
video_path = "" #@param {type:"string"}
jumlah_klip = 5 #@param {type:"integer"}
durasi_per_klip = 45 #@param {type:"integer"}
model_whisper = "base" #@param ["tiny", "base", "small", "medium"]
bahasa = "en" #@param ["en", "id", "auto"]

# --- SETUP MODEL WHISPER ---
print("📥 Memuat model AI Whisper...")
model = whisper.load_model(model_whisper)

def get_duration(filename):
    cmd = f'ffprobe -v error -show_entries format=duration -of default=noprint_wrappers=1:nokey=1 "{filename}"'
    return float(subprocess.check_output(cmd, shell=True))

def run_ffmpeg(cmd):
    result = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    if result.returncode != 0:
        raise RuntimeError(f"FFmpeg error:\n{result.stderr}")

# --- VALIDASI FILE ---
if not os.path.exists(video_path):
    raise FileNotFoundError(f"Video tidak ditemukan: {video_path}")

file_name_clean = os.path.splitext(os.path.basename(video_path))[0]
folder_path = os.path.dirname(video_path)

total_sec = get_duration(video_path)
interval = total_sec / jumlah_klip

if jumlah_klip * durasi_per_klip > total_sec:
    print(f"⚠️ Peringatan: total durasi klip ({jumlah_klip * durasi_per_klip}s) melebihi durasi video ({int(total_sec)}s). Klip mungkin overlap.")

print(f"🎬 Memulai pemrosesan untuk: {file_name_clean}\n")

# --- LOOP UTAMA ---
for i in tqdm(range(jumlah_klip), desc="📦 Total Progress", unit="klip"):
    start_time = min(i * interval, total_sec - durasi_per_klip)
    base_output_name = f"{file_name_clean}_Klip_{i+1}"

    output_clip = os.path.join(folder_path, f"{base_output_name}.mp4")
    transcript_file = os.path.join(folder_path, f"{base_output_name}.txt")

    tqdm.write(f"\n--- 🔄 Memproses Bagian {i+1} (Mulai: {int(start_time)}s) ---")

    # LANGKAH 1: Potong Video
    with tqdm(total=1, desc="  ✂️  Memotong video", leave=False) as pbar:
        cut_cmd = f'ffmpeg -ss {start_time} -t {durasi_per_klip} -i "{video_path}" -c copy "{output_clip}" -y'
        run_ffmpeg(cut_cmd)
        pbar.update(1)

    # LANGKAH 2: Transkripsi Whisper
    with tqdm(total=1, desc="  📝 Transkripsi audio", leave=False) as pbar:
        lang_param = None if bahasa == "auto" else bahasa
        result = model.transcribe(output_clip, language=lang_param)
        with open(transcript_file, "w", encoding="utf-8") as f:
            f.write(result["text"].strip())
        pbar.update(1)

    # LANGKAH 3: UBAH MENJADI 9:16
    output_916 = os.path.join(folder_path, f"{base_output_name}_916.mp4")
    with tqdm(total=1, desc="  📐 Resize ke 9:16  ", leave=False) as pbar:
        resize_cmd = (
            f'ffmpeg -i "{output_clip}" '
            f'-vf "split[fg][bg]; '
            f'[bg]scale=1080:1920:force_original_aspect_ratio=increase,'
            f'crop=1080:1920,boxblur=40:40[blurred]; '
            f'[fg]scale=1080:1920:force_original_aspect_ratio=decrease[small]; '
            f'[blurred][small]overlay=(W-w)/2:(H-h)/2" '
            f'-c:a copy "{output_916}" -y'
        )
        run_ffmpeg(resize_cmd)
        pbar.update(1)

    tqdm.write(f"  ✅ Selesai: {base_output_name}_916.mp4 & .txt")

print("\n\n🚀 SEMUA PROSES SELESAI!")
```

```python
#@title 🎙️ Transkrip Audio dari Google Drive
#@markdown Masukkan path video/audio dari Google Drive dan jalankan sel ini.

# Instal dependensi
import subprocess
import sys

subprocess.run([sys.executable, "-m", "pip", "install", "openai-whisper"], check=True)

import whisper
import os
from tqdm.notebook import tqdm
from google.colab import drive

# --- MOUNT GOOGLE DRIVE ---
print("📂 Menghubungkan ke Google Drive...")
drive.mount('/content/drive')

# --- KONFIGURASI PARAMETER ---
#@markdown ---
#@markdown ### Konfigurasi
input_path = "" #@param {type:"string"}
#@markdown > Bisa berupa **file tunggal** (`.mp4`, `.mp3`, dll) atau **folder** berisi banyak file
model_whisper = "base" #@param ["tiny", "base", "small", "medium"]
bahasa = "en" #@param ["en", "id", "auto"]
simpan_ke_drive = True #@param {type:"boolean"}
#@markdown > Jika `True`, file `.txt` disimpan di folder yang sama dengan file input

# --- EKSTENSI YANG DIDUKUNG ---
SUPPORTED_EXT = {".mp4", ".mkv", ".avi", ".mov", ".mp3", ".wav", ".m4a", ".flac", ".ogg"}

# --- KUMPULKAN DAFTAR FILE ---
def collect_files(path):
    files = []
    if os.path.isfile(path):
        ext = os.path.splitext(path)[1].lower()
        if ext in SUPPORTED_EXT:
            files.append(path)
        else:
            print(f"⚠️ Format tidak didukung: {ext}")
    elif os.path.isdir(path):
        for fname in sorted(os.listdir(path)):
            ext = os.path.splitext(fname)[1].lower()
            if ext in SUPPORTED_EXT:
                files.append(os.path.join(path, fname))
        print(f"📁 Ditemukan {len(files)} file media dalam folder.")
    else:
        raise FileNotFoundError(f"Path tidak ditemukan: {path}")
    return files

# --- SETUP MODEL WHISPER ---
print(f"\n📥 Memuat model Whisper '{model_whisper}'...")
model = whisper.load_model(model_whisper)
print("✅ Model siap.\n")

# --- PROSES TRANSKRIPSI ---
file_list = collect_files(input_path)

if not file_list:
    print("❌ Tidak ada file media yang ditemukan.")
else:
    lang_param = None if bahasa == "auto" else bahasa
    hasil_ringkasan = []

    for idx, file_path in enumerate(tqdm(file_list, desc="📦 Total Progress", unit="file")):
        file_name = os.path.basename(file_path)
        file_stem = os.path.splitext(file_name)[0]

        tqdm.write(f"\n--- 🔄 [{idx+1}/{len(file_list)}] Memproses: {file_name} ---")

        try:
            # TRANSKRIPSI
            with tqdm(total=1, desc="  📝 Transkripsi audio", leave=False) as pbar:
                result = model.transcribe(file_path, language=lang_param)
                teks = result["text"].strip()
                pbar.update(1)

            # TENTUKAN LOKASI SIMPAN
            if simpan_ke_drive:
                output_dir = os.path.dirname(file_path)
            else:
                output_dir = "/content"

            transcript_path = os.path.join(output_dir, f"{file_stem}_transkrip.txt")

            # SIMPAN FILE TEKS
            with open(transcript_path, "w", encoding="utf-8") as f:
                f.write(teks)

            tqdm.write(f"  ✅ Transkrip disimpan: {transcript_path}")
            tqdm.write(f"  📄 Pratinjau: {teks[:120]}{'...' if len(teks) > 120 else ''}")
            hasil_ringkasan.append(("✅", file_name, transcript_path))

        except Exception as e:
            tqdm.write(f"  ❌ Gagal memproses {file_name}: {e}")
            hasil_ringkasan.append(("❌", file_name, str(e)))

    # --- RINGKASAN AKHIR ---
    print("\n" + "="*55)
    print("📊 RINGKASAN HASIL TRANSKRIPSI")
    print("="*55)
    for status, nama, info in hasil_ringkasan:
        print(f"  {status}  {nama}")
        print(f"      → {info}")
    print("="*55)
    berhasil = sum(1 for s, _, _ in hasil_ringkasan if s == "✅")
    print(f"\n🚀 Selesai! {berhasil}/{len(file_list)} file berhasil ditranskrip.")
```

```python
import os
import google_auth_oauthlib.flow
import googleapiclient.discovery
import googleapiclient.errors
from googleapiclient.http import MediaFileUpload
from google.colab import drive # Import drive

CLIENT_SECRETS_FILE = "path/to/your/client_secret.json" # @param {type:"string"}
VIDEO_FILE = "path/to/your/video.mp4" # @param {type:"string"}
PRIVACY_STATUS = "unlisted" # @param ["private", "public", "unlisted"]
CATEGORY_ID = "22" # @param ["1", "10", "17", "20", "22", "27", "28"]

scopes = ["https://www.googleapis.com/auth/youtube.upload"]

def upload_video():
    # Pastikan Google Drive terpasang jika file kredensial atau video ada di Drive
    if not os.path.exists('/content/drive'):
        print("Menghubungkan Google Drive...")
        drive.mount('/content/drive')
    else:
        print("Google Drive sudah terhubung.")

    # Validasi File
    if not os.path.exists(CLIENT_SECRETS_FILE):
        print(f"❌ ERROR: File JSON tidak ditemukan di {CLIENT_SECRETS_FILE}")
        return
    if not os.path.exists(VIDEO_FILE):
        print(f"❌ ERROR: File Video tidak ditemukan di {VIDEO_FILE}")
        return

    print("🔐 Memulai proses autentikasi...")
    flow = google_auth_oauthlib.flow.InstalledAppFlow.from_client_secrets_file(
        CLIENT_SECRETS_FILE, scopes)

    # Kita ubah redirect_uri-nya ke alamat khusus Google agar tidak ke localhost
    flow.redirect_uri = 'https://localhost'

    auth_url, _ = flow.authorization_url(prompt='consent')

    print(f"\n1. Buka link ini di browser:\n{auth_url}")
    print("\n2. Login, klik 'Advanced' -> 'Go to... (unsafe)', lalu klik 'Allow'.")
    print("3. Setelah itu, browser akan error/kosong. JANGAN PANIK!")
    print("4. COPY (Salin) seluruh URL yang ada di address bar browser tersebut.")

    # Input manual URL yang kamu dapatkan tadi
    code_url = input("\n5. TEMPEL (Paste) URL hasil redirect tadi di sini: ")

    # Script akan mengambil kode otomatis dari URL tersebut
    import urllib.parse as urlparse
    parsed = urlparse.urlparse(code_url)
    auth_code = urlparse.parse_qs(parsed.query)['code'][0]

    flow.fetch_token(code=auth_code)
    credentials = flow.credentials
    # ------------------------------------------
    youtube = googleapiclient.discovery.build("youtube", "v3", credentials=credentials)

    # Ambil nama file video dari path VIDEO_FILE (tanpa ekstensi)
    video_title = os.path.splitext(os.path.basename(VIDEO_FILE))[0]

    request_body = {
        "snippet": {
            "categoryId": CATEGORY_ID,
            "description": "Video diunggah otomatis dari Google Drive via Python",
            "title": video_title # Menggunakan nama file sebagai judul
        },
        "status": {
            "privacyStatus": PRIVACY_STATUS
        }
    }

    # Proses Upload
    print("Sedang mengunggah video...")
    insert_request = youtube.videos().insert(
        part="snippet,status",
        body=request_body,
        media_body=MediaFileUpload(VIDEO_FILE, chunksize=-1, resumable=True)
    )

    response = insert_request.execute()
    print(f"Selesai! Video berhasil diunggah. ID Video: {response['id']}")
    print(f"Link: https://www.youtube.com/watch?v={response['id']}")

if __name__ == "__main__":
    upload_video()
```
